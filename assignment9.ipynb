```python
# Task 2.2. Tuples: ordered, immutable collections with duplicates.

coordinates = (10, 20, 10)
print("Original tuple:", coordinates)
print("First element:", coordinates[0])
print("Slice:", coordinates[:2])
print("Count of 10:", coordinates.count(10))
print("Index of 20:", coordinates.index(20))

x, y, z = coordinates
print("Unpacked values:", x, y, z)

single_item = (5,)
print("Single-item tuple:", single_item)
print("Its type:", type(single_item))

combined = coordinates + (30, 40)
print("Combined tuple:", combined)

try:
    coordinates[0] = 99
except TypeError as error:
    print("Cannot change a tuple element:", error)

converted = list(coordinates)
converted[0] = 99
print("Modified list:", converted)
print("Converted back to tuple:", tuple(converted))
print("Original tuple:", coordinates)
```

```python
# Task 2.3. Dictionaries: mutable collections of key-value pairs.

student = {
    "name": "Alex",
    "programme": "MiBA",
    "grade": 85
}

print("Original dictionary:", student)
print("Name:", student["name"])
print("Missing key with a default:", student.get("age", "Not specified"))

student["grade"] = 90
student["city"] = "Saint Petersburg"
print("After updating and adding values:", student)

student.update({"grade": 95, "year": 1})
print("After update:", student)

print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))

removed_city = student.pop("city")
print("Removed city:", removed_city)

print("Contains the key 'grade':", "grade" in student)

for key, value in student.items():
    print(f"{key}: {value}")

duplicate_keys = {"name": "Alex", "name": "Maria"}
print("Duplicate keys keep the last value:", duplicate_keys)
```

```python
# Task 2.4. Sets: collections of unique elements without positional indexing.

python_students = {"Alex", "Maria", "Ivan", "Alex"}
sql_students = {"Maria", "Ivan", "Anna"}

# Sorting is used only to make printed results easy to read.
print("Unique Python students:", sorted(python_students))
print("SQL students:", sorted(sql_students))

print("Union:", sorted(python_students | sql_students))
print("Intersection:", sorted(python_students & sql_students))
print("Only Python:", sorted(python_students - sql_students))
print("Only one course:", sorted(python_students ^ sql_students))

python_students.add("Olga")
print("After add:", sorted(python_students))

python_students.update(["Pavel", "Anna"])
print("After update:", sorted(python_students))

python_students.remove("Alex")
print("After remove:", sorted(python_students))

python_students.discard("Unknown")
print("Discarding a missing element does not cause an error.")

print("Maria is a Python student:", "Maria" in python_students)
print("Subset check:", {"Maria", "Ivan"}.issubset(python_students))
print("Disjoint check:", python_students.isdisjoint({"Max"}))

try:
    python_students.add(["New", "Student"])
except TypeError as error:
    print("A set cannot contain a list:", error)

empty_set = set()
print("Empty set type:", type(empty_set))
print("Empty braces type:", type({}))
```

```python
# Task 2.5. Nested collections.

# A list of lists.
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

print("Matrix:", matrix)
print("Second row, third element:", matrix[1][2])

matrix[0][1] = 99
print("Updated matrix:", matrix)

# A list of dictionaries containing lists.
students = [
    {"name": "Alex", "grades": [80, 90, 85]},
    {"name": "Maria", "grades": [95, 88, 92]}
]

students[0]["grades"].append(100)

for student in students:
    average = sum(student["grades"]) / len(student["grades"])
    print(f"{student['name']}: average grade = {average:.2f}")

# A dictionary containing another dictionary and a set.
course = {
    "title": "Python",
    "teacher": {"name": "Dr Smith", "office": 207},
    "topics": {"strings", "lists", "dictionaries"}
}

print("Teacher's office:", course["teacher"]["office"])

course["topics"].add("sets")
print("Course topics:", sorted(course["topics"]))

# A tuple can contain a mutable list.
record = ("Alex", [80, 90])
record[1].append(100)
print("Tuple after changing the list inside it:", record)
```

```python
# Task 2.6. Assignment, shallow copy and deep copy.

import copy

original = [[1, 2], [3, 4]]

alias = original
shallow_copy = original.copy()
deep_copy = copy.deepcopy(original)

original[0].append(99)

print("After changing a nested list:")
print("Original:", original)
print("Alias:", alias)
print("Shallow copy:", shallow_copy)
print("Deep copy:", deep_copy)

original.append([5, 6])

print("\nAfter adding a new outer element:")
print("Original:", original)
print("Alias:", alias)
print("Shallow copy:", shallow_copy)
print("Deep copy:", deep_copy)

print("\nAlias is the original object:", alias is original)
print("Shallow copy is the original object:", shallow_copy is original)
print("Shallow copy shares the first inner list:",
      shallow_copy[0] is original[0])
print("Deep copy shares the first inner list:",
      deep_copy[0] is original[0])
```
